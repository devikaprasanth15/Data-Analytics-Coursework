```python
## Python Assignment 3: While Loop, For loop and Function

## While loop & Control Statements: (else, break, continue, pass)

## Number Guessing Game

import random

# Generate a random number between 1 and 10:
secret_number = random.randint(1, 10)

# Set maximum number of attempts:
attempt = 3

while attempt >0:
    guess_number = int(input("Guess the number between 1 and 10: "))
    print ("The number you guessed is:", guess_number)

    if guess_number < 1 or guess_number > 10:
        print ("Your guess is out of range! Please guess a number between 1 and 10")
        continue

    attempt -=1

    if guess_number == secret_number:
        print("Congratulations! You guessed the correct number!")
        break
    elif guess_number < secret_number:
        print("Too low. Try again.")
    else:
        print("Too high. Try again.")

else:
    print("Better luck next time!")
```

```python
## Function:

## BMI Calculator

# Define function to calculate BMI:
def calculate_bmi(weight, height):
    bmi = weight / (height ** 2)
    return bmi

# Ask user for weight and height:
weight = float(input("Enter your weight in kg: "))
print ("Your Weight:" ,weight)
height = float(input("Enter your height in meters: "))
print ("Your Height:", height)

# Call the function:
bmi = calculate_bmi(weight, height)

# Display the result
print("Your BMI is:" , format(bmi, ".2f"))

## BMI Category:
if bmi < 18.5:
    print("Category: Underweight")
elif bmi < 25:
    print("Category: Healthy Weight (Normal)")
elif bmi < 30:
    print("Category: Overweight")
else:
    print("Category: Obesity")
```
